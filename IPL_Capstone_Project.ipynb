## Key Player Performances

1. Most Player of the Match Awards

```python
count1 = df['player_of_the_match'].value_counts().head(10)
count1
```

```python
sns.barplot(x = count1.values, y = count1.index, palette = 'viridis')
plt.title('Players with most MOTM')
```

2. Top Scorers

```python
high = df.groupby('top_scorer')['highscore'].sum().sort_values(ascending = False).head()
high
```

```python
sns.barplot(x = high.values, y = high.index, palette = 'viridis')
```

3. Ten Best Bowling Figures

```python
df.head()
```

```python
df['highest_wicket'] = df['best_bowling_figure'].apply(lambda x : x.split('--')[0])
df['highest_wicket'] = df['highest_wicket'].astype(int)
top_bowlers = df.groupby('best_bowling')['highest_wicket'].sum().sort_values(ascending = False).head(10)
top_bowlers
```

```python
sns.barplot(x = top_bowlers.values, y = top_bowlers.index, palette = 'viridis')
plt.title('Bowlers with top 10 best bowling figures')
```

4. Venue Counts

```python
venue_count = df['venue'].value_counts()
venue_count
```

```python
sns.barplot(x = venue_count.values, y = venue_count.index, palette = 'viridis')
plt.xlabel('No. of Matches')
```
